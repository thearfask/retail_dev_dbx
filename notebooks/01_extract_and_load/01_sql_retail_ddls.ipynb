```sql
CREATE TABLE IF NOT EXISTS retail_dev_schema.bronze_users (
    id BIGINT,
    first_name STRING,
    last_name STRING,
    email STRING,
    country STRING,
    batch_id STRING,
    ingested_at TIMESTAMP
)
USING DELTA;


CREATE TABLE IF NOT EXISTS retail_dev_schema.bronze_products (
    id BIGINT,
    title STRING,
    category STRING,
    brand STRING,
    price DECIMAL(18,2),
    discount_percentage DECIMAL(10,2),
    stock BIGINT,
    batch_id STRING,
    ingested_at TIMESTAMP
)
USING DELTA;


CREATE TABLE IF NOT EXISTS retail_dev_schema.bronze_carts (
    id BIGINT,
    user_id BIGINT,
    products ARRAY<STRUCT<
        id: BIGINT,
        title: STRING,
        price: DECIMAL(18,2),
        quantity: BIGINT,
        total: DECIMAL(18,2),
        discountPercentage: DECIMAL(10,2),
        discountedTotal: DECIMAL(18,2),
        thumbnail: STRING
    >>,
    total DECIMAL(18,2),
    discounted_total DECIMAL(18,2),
    total_products BIGINT,
    total_quantity BIGINT,
    batch_id STRING,
    ingested_at TIMESTAMP
)
USING DELTA;
```
